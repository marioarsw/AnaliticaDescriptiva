# Preguntas descriptivas

### 1. Distribución de resultados de laboratorio

```python
# 1. Distribución de resultados de laboratorio

df['RESULTADO_PCR'].value_counts(dropna=False)
```

```python
# Para verlo en proporciones
df['RESULTADO_PCR'].value_counts(normalize=True) * 100
```

# Preguntas descriptivas

### 2. Distribución de casos positivos por grupo de edad

```python
bins = [0, 17, 29, 44, 59, 74, 120]
labels = ['0-17', '18-29', '30-44', '45-59', '60-74', '75+']

df['GRUPO_EDAD'] = pd.cut(df['EDAD'], bins=bins, labels=labels)

positivos = df[df['RESULTADO_PCR'] == 34]

positivos['GRUPO_EDAD'].value_counts().sort_index()
```

```python
# en porcentajes

positivos['GRUPO_EDAD'].value_counts(normalize=True).sort_index() * 100
```

# Preguntas descriptivas

3. Prevalencia de comorbilidades en casos positivos

```python
positivos = df[df['RESULTADO_PCR'] == 34]

for col in ['NEUMONIA','ASMA','DIABETES', 'HIPERTENSION', 'OBESIDAD', 'TABAQUISMO']:
    porcentaje = (positivos[col] == 1).mean() * 100
    print(f"{col}: {porcentaje:.2f}%")
```

# Preguntas descriptivas

4. Proporción de casos que requirieron UCI

```python
# Personas que entrar a los cuidados intensivos
proporcion_uci = (df['UCI'] == 1).mean() * 100
proporcion_uci
```
