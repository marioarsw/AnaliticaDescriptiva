# Generators
Many objects in Python support iteration, such as over objects in a list or lines in a file. This is accomplished by means of the iterator protocol, a generic way to make objects iterable. For example, iterating over a dictionary yields the dictionary keys:

Los generadores se usan para recorrer una lista (por ejemplo) a través de la palabra reservada yield, para recorrer un elemento se tiene que usar la palabra next()

```python
def contador(n):
    print("Generator iniciado")
    for i in range(n):
        print(f"Generando {i}")
        yield i
    print("Generator terminado")
```

```python
g = contador(5)
print(g)
```

```python
next(g)
```

```python
for x in g:
    print("Recibido:", x)
```
