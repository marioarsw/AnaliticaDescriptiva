* Objetivo
Utilizar GridSpec para organizar subplots de diferentes tamaños en un análisis de ventas.

* Instrucciones:
* Genera un DataFrame de ventas con Mes, Ventas, y Gastos.
* Utiliza randint para generar números aleatorios enteros. Para ventas los valores deben estar entre 2000 y 5000. Para gastos, los valores debn estar entre 1000 y 4000.
* Usa GridSpec para organizar los subplots de la siguiente manera:
    * Primer gráfico (barra, ocupando dos columnas)
    * Segundo gráfico (línea, una columna)
    * Tercer gráfico (dispersión, dos filas).
    * Agrega títulos y ajusta espacios.

```python
import seaborn as sns
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec
import pandas as pd
import numpy as np
```

* Genera un DataFrame de ventas con Mes, Ventas, y Gastos.

* Utiliza randint para generar números aleatorios enteros. Para ventas los valores deben estar entre 2000 y 5000. Para gastos, los valores deben estar entre 1000 y 4000.

```python
# Meses
meses = [
    'Enero', 'Febrero', 'Marzo', 'Abril',
    'Mayo', 'Junio', 'Julio', 'Agosto',
    'Septiembre', 'Octubre', 'Noviembre', 'Diciembre'
]

# Datos aleatorios
ventas = np.random.randint(2000, 5001, size=12)
gastos = np.random.randint(1000, 4001, size=12)

# DataFrame
df_ventas = pd.DataFrame({
    'Mes': meses,
    'Ventas': ventas,
    'Gastos': gastos
})

df_ventas
```

* Usa GridSpec para organizar los subplots de la siguiente manera:
    * Primer gráfico (barra, ocupando dos columnas)
    * Segundo gráfico (línea, una columna)
    * Tercer gráfico (dispersión, dos filas).
    * Agrega títulos y ajusta espacios.

```python
fig = plt.figure(figsize=(18, 10))
gs = GridSpec(2, 3, figure=fig)   # 2 filas 3 columnas

ax1 = fig.add_subplot(gs[0, :2])  # dos columnas
ax2 = fig.add_subplot(gs[0, 2])   # una columna
ax3 = fig.add_subplot(gs[1, :])   # dos filas (0,1)

ax1.bar(df_ventas['Mes'], df_ventas['Ventas'], color='#4C72B0', alpha=0.7)
ax1.set_title('Ventas Mensuales')
ax1.set_xlabel('Mes')
ax1.set_ylabel('Ventas')

ax2.plot(df_ventas['Mes'], df_ventas['Gastos'], marker='o', color='#C44E52',  alpha=0.7)
ax2.set_title('Gastos Mensuales')
ax2.set_xlabel('Mes')
ax2.set_ylabel('Gastos')

ax3.scatter(df_ventas['Ventas'], df_ventas['Gastos'], color='#55A868', alpha=0.7)
ax3.set_title('Relación entre Ventas y Gastos')
ax3.set_xlabel('Ventas')
ax3.set_ylabel('Gastos')

plt.tight_layout()
plt.show()
```
