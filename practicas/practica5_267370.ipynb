Crea un arreglo unidimensional con 15 números aleatorios entre 1 y 50. Luego:

1. Selecciona los elementos mayores que 25 utilizando una máscara booleana.
2. Cuenta cuántos elementos son mayores que 25.

Imporación de librerías.

```python
import numpy as np
```

Crea un arreglo unidimensional con 15 números aleatorios entre 1 y 50.

```python
A = np.random.randint(1, 51, 15)

A
```

1. Selecciona los elementos mayores que 25 utilizando una máscara booleana.

```python
print(A[A > 25])
```

2.
Cuenta cuántos elementos son mayores que 25.

```python
print((A > 25).sum())
```
