### Distribuciones
Una distribución describe cómo se reparten los valores de una variable y con qué frecuencia aparecen.

- Normal: simétrica, forma de campana (media y desviación estándar)
- Uniforme: todos los valores tienen la misma probabilidad
- Exponencial: muchos valores pequeños, pocos grandes
- Poisson: distribución discreta para conteo de eventos

NumPy permite generar datos que siguen estas distribuciones para simular datos y analizarlos.


### 🔵 1. Distribución Normal (campana)

🧠 Idea mental: la famosa campana de Gauss

* La mayoría de los valores están cerca del centro

* Pocos valores extremos

* Es simétrica

Ejemplos reales:

* Estatura

* Calificaciones

* Errores de medición

```
# np.random.normal(loc=0, scale=1, size=1000)
```
Qué significan los parámetros:

loc → media (centro)

scale → desviación estándar (qué tan ancha es la campana)

size → cuántos datos generas

### 🟢 2. Distribución Uniforme

🧠 Idea mental: todos los valores tienen la misma probabilidad

* No hay valores más comunes que otros

* El histograma se ve “plano”

Ejemplos:

* Número aleatorio entre 0 y 1

* Sorteos justos

En NumPy:

```
# np.random.uniform(low=0, high=1, size=1000)
```

Parámetros:

low → mínimo

high → máximo

size → cantidad de datos

### 🟠 3. Distribución Exponencial

🧠 Idea mental: muchos valores pequeños, pocos grandes

* No es simétrica

* Está sesgada a la derecha

Ejemplos:

* Tiempo de espera

* Vida útil de componentes

* Tiempo hasta que ocurre un evento

En NumPy:

```
# np.random.exponential(scale=1, size=1000)
```
Parámetro:

scale → valor promedio aproximado

### 🔴 4. Distribución Poisson

🧠 Idea mental: cuenta eventos

* Valores enteros (0, 1, 2, 3…)

* No continuos

* Muy usada para conteos

Ejemplos:

* Llamadas por hora

* Clientes que llegan a una tienda

* Errores por página

En NumPy:

```
# np.random.poisson(lam=5, size=1000)
```

Parámetro clave:

lam → número promedio de eventos



1. Genera un DataFrame con Pandas que contenga 1000 valores en cuatro columnas (Normal, Uniforme, Exponencial, Poisson).

````python
data = {
    'Normal': np.random.normal(loc=0, scale=1, size=1000),
    'Uniforme': np.random.uniform(low=0, high=1, size=1000),
    'Exponencial': np.random.exponential(scale=1, size=1000),
    'Poisson': np.random.poisson(lam=5, size=1000)
}

df = pd.DataFrame(data)

df.head()
````

2. Crea una figura con plt.subplots(2, 2) para organizar los histogramas en una cuadrícula de 2x2.

````python
fig, ax = plt.subplots(2, 2, figsize=(10, 8))

# Para agregar color edgecolor='black'

# Normal
ax[0, 0].hist(df['Normal'], bins=30, color='#4C72B0', alpha=0.7)
ax[0, 0].set_title('Distribución Normal')
ax[0, 0].set_xlabel('Valor')
ax[0, 0].set_ylabel('Frecuencia')

# Uniforme
ax[0, 1].hist(df['Uniforme'], bins=30, color='#55A868', alpha=0.7)
ax[0, 1].set_title('Distribución Uniforme')
ax[0, 1].set_xlabel('Valor')
ax[0, 1].set_ylabel('Frecuencia')

# Exponencial
ax[1, 0].hist(df['Exponencial'], bins=30, color='#C44E52', alpha=0.7)
ax[1, 0].set_title('Distribución Exponencial')
ax[1, 0].set_xlabel('Valor')
ax[1, 0].set_ylabel('Frecuencia')

# Poisson
ax[1, 1].hist(df['Poisson'], bins=30, color='#8172B3', alpha=0.7)
ax[1, 1].set_title('Distribución Poisson')
ax[1, 1].set_xlabel('Valor')
ax[1, 1].set_ylabel('Frecuencia')

plt.tight_layout()
plt.show()
````
