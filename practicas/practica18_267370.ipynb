# Ejercicio 02: Extraer componentes de fecha y tiempo

1. De la columna 'DateTime' extrae el año, mes, día y hora. Crea una nueva columna para cada componente que extraes.
2. Imprime las primeras tres líneas del dataframe que ahora deberá tener 4 nuevas columnas.

```python
airdata['year'] = airdata['DateTime'].dt.year
airdata['month'] = airdata['DateTime'].dt.month
airdata['day'] = airdata['DateTime'].dt.day
airdata['hour'] = airdata['DateTime'].dt.hour
```

# Ejercicio 03: Aritmética - Suma de Intervalos (timedelta)

1. Suma 10 días, 7 horas y 15 minutos a cada elemento de la columna 'DateTime'. 2. Guarda el resultado en una nueva columna y agrega al dataframe.
3. Imprime la columna 'DateTime' y la columna que resulta de sumar 10 días, 7 horas y 15 minutos.

```python
airdata['more_time'] = airdata['DateTime'] + pd.Timedelta(days=10, hours=7, minutes=15)
```

```python
airdata[['DateTime', 'more_time']]
```

# Ejercicio 04: Timestamp

1. Convierte la columna 'DateTime' la cual tiene un formato de fecha legible a a timestamp Unix. Guarda el resultado en una nueva columna.
2. Imprime las primeras tres filas de  la columna 'DateTime' y la columna que representa dicha fecha en timestamp Unix.

```python
airdata['timestamp_unix'] = pd.to_datetime(airdata['DateTime']).astype('int64') // 10**9

# astype('int64')
# convertir la fecha a número de nanosegundos desde 1970.

# // 10**9
# dividir entre 1,000,000,000 para convertir nanosegundos → segundo
```

```python
print(airdata[['DateTime', 'timestamp_unix']].head(3))
```
