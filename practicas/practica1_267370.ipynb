### Lectura de archivo y explicación de algoritmo

1. Se declara una lista vacía
2. Se recorre el archivo
3. Se limpian los carácteres vacíos a la derecha
4. Se usa 'starstwith'para encontrar coincidencias con 'X-DSPAM-Confidence: '
5. A los valores encontrados, se aplica 'split' y accede al valor [1] para traer únicamente los valores numéricos
6. Se usa 'append' para agregar los valores (casteados a float) a la lista declarada al inicio

```python
archivo = open('/content/drive/MyDrive/ClassFiles/mbox.txt')

valores = [] # Lista

for linea in archivo:
  linea = linea.rstrip() # Limpiar a la derecha
  if linea.startswith('X-DSPAM-Confidence: '): # Encontrar coincidencias
    valor = linea.split(": ")[1] # Splitear y acceder al valor
    valores.append(float(valor)) # append -> agregar

```

### Mostrar valores

```python
print(valores)
```

### Cantidad de datos extraídos

```python
print(len(valores))

# o bien, contar las líneas ya que equivalen a un valor

count = 0
for linea in valores:
  count = count + 1
print(count)
```

### Sumatoria de los datos

```python
print(sum(valores))
```

```python
# De manera manual
sumatoria_valores = 0

for x in valores:
  sumatoria_valores = sumatoria_valores + x

print(sumatoria_valores)
```

### Promedio de los datos

```python
print(sum(valores)/len(valores))
```

### Calcular la varianza

La varianza calculada mide la dispersión de los valores de X-DSPAM-Confidence (spliteados anteriormente) respecto a su promedio.

Para calcular la varianza, se puede hacer con la librería statistics de python o también utilizando numpy, para esta práctica elegí statistics

```python
import statistics
print(statistics.pvariance(valores))
```
